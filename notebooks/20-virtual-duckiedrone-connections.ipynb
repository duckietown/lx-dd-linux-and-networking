# Virtual Duckiedrone Connections

A virtual Duckiedrone runs locally on the base station. This notebook introduces the Duckietown Shell virtual-device workflow.

## Create a virtual Duckiedrone

Run this command from a base-station terminal where the Duckietown Shell is installed.

Create one virtual Duckiedrone once. In this command, `-t duckiedrone` selects the Duckiedrone type, literal `-c DD24` selects the DD24 configuration, and `DUCKIEDRONE_NAME` sets its hostname:

```bash
dts duckiebot virtual create -t duckiedrone -c DD24 DUCKIEDRONE_NAME
```

This creates a virtual Duckiedrone DD24 on the base station. Unlike SSH, connecting to it does not establish a network connection because the virtual device is already running locally.

## List, start, and connect

For an existing virtual Duckiedrone, inspect its status, start it, and open a shell:

```bash
dts duckiebot virtual list
dts duckiebot virtual start DUCKIEDRONE_NAME
dts duckiebot virtual connect DUCKIEDRONE_NAME
```

`dts duckiebot virtual connect` checks for the requested virtual Duckiedrone and opens an interactive Bash shell inside it. Use this command in this LX; implementation details of its local runtime are outside the scope of this lesson.

Use `exit` when finished to return to the base-station shell.

## Local virtual and SSH connections

A physical Duckiedrone needs an authorized network route, identity verification, authentication, and SSH. A virtual Duckiedrone connection opens a local virtual-device shell, so `ssh-copy-id` is not part of this workflow. Network traffic produced by the virtual device still leaves through the base station and remains subject to the applicable network policy.
