# Shell Scripts and Execution

A __script__ is a text file containing commands. Work only with scripts that you have read and trust, because running or sourcing a script executes its commands.

## Create and run a trusted script

Create a script in a temporary practice area, inspect it, make it executable, and run it:

```bash
PRACTICE_DIR=/tmp/linux-script-practice
mkdir -p "$PRACTICE_DIR"
cd "$PRACTICE_DIR"
printf '%s\n' '#!/usr/bin/env bash' > greeting.sh
printf '%s\n' 'printf "%s\n" "Hello from a script"' >> greeting.sh
cat greeting.sh
chmod u+x greeting.sh
./greeting.sh
```

The first line is a [__shebang__](https://en.wikipedia.org/wiki/Shebang_(Unix)) that tells Linux to use Bash. `chmod u+x` gives the file owner permission to execute it. The `./` prefix means run the file from this directory; the shell does not normally search the current directory for commands. Inspect the permissions with `ls -l greeting.sh`: `x` means executable, while `r` and `w` mean readable and writable.

## Run versus source

`bash SCRIPT_NAME` and `./SCRIPT_NAME` start a separate process. The script can read exported environment variables, but assignments it makes do not change your current shell. `source SCRIPT_NAME` runs the file in the current shell, so assignments and directory changes remain afterward:

```bash
unset LX_MESSAGE
printf '%s\n' 'LX_MESSAGE="Shell setup complete"' > trusted_environment.sh
cat trusted_environment.sh
bash trusted_environment.sh
printf 'After bash: %s\n' "${LX_MESSAGE:-not set}"
source ./trusted_environment.sh
printf 'After source: %s\n' "$LX_MESSAGE"
```

`unset LX_MESSAGE` clears a value left by an earlier run. The first result should say `not set`, because Bash used a separate process. The second should say `Shell setup complete`, because `source` ran the assignment in the current shell. Setup scripts can define environment values this way.

Source only a script that you have read and trust. Sourcing acts as though you typed its commands into your current shell, so an untrusted script can change variables, change directories, or run commands you did not intend.
