# Shell Help, Errors, and Exit Status

A shell keeps track of and can provide evidence about what just happened. 

This notebook covers built-in help, recovery from common errors, and command exit status.

## Get help and control commands

Use `--help` for a short description of a command's options:

```bash
ls --help
```

On a full Ubuntu system, `man ls` opens a local manual page for `ls`; press `q` to leave it. The minimized LX image does not include manual-page content, so use `--help` here and the external references below. 

## Shell must-know keyboard shortcuts

A few fundamental functionalities:

- Press `Tab` after part of a filename to request completion. 
- The Up and Down arrow keys recall earlier commands. 
- Press `Ctrl-C` to interrupt a command that is still running; it does not undo a completed command.
- `Ctrl-Shift-C` to copy, `Ctrl+Shift-V` to paste 
- `Ctrl-R`: reverse command search - start typing a command, and autocomplete based on priori input history.

## Recover from common errors

Errors are useful evidence. Read the first line, then check the command's location and spelling instead of adding `sudo` or repeatedly rerunning it:

- For `No such file or directory`, run `pwd` and `ls` to confirm the current directory and exact filename.

- For `command not found`, check spelling and run `command -v COMMAND_NAME` to see whether the shell can find the command.

- For `Permission denied` when running a script, inspect it with `ls -l filename`; the file may need execute permission.

## Check command results

Every command finishes with an exit status. `0` normally means success. A nonzero status reports a command-specific result, which can be an error or an expected condition. `$?` is the status of the command that just finished, so inspect it immediately. From the LX root, where `ls packages` succeeds, run:

```bash
python3 -m py_compile packages/shell_exercises/*.py
printf 'Exit status: %s\n' "$?"
```

Use `&&` when the next command should run only after the previous command succeeds:

```bash
python3 -m py_compile packages/shell_exercises/*.py && printf '%s\n' 'Python files are syntactically valid'
```

Use `||` when the next command should run only after the previous command returns a nonzero status. This self-contained example intentionally searches for absent text:

```bash
PRACTICE_DIR=/tmp/linux-status-practice
mkdir -p "$PRACTICE_DIR"
printf 'banana\napple\npear\n' > "$PRACTICE_DIR/fruit.txt"
grep -n 'orange' "$PRACTICE_DIR/fruit.txt" || printf '%s\n' 'No matching fruit found'
```

A missing `orange` line is expected here, so the `printf` message handles `grep`'s nonzero no-match result. A command that prints nothing may still have succeeded; use its exit status and the files it produced to decide what happened.
