## Create and inspect files

`printf` prints text. The `>` operator redirects that output into `fruit.txt`: it creates the file when it does not exist, or replaces its contents when it does:

```bash
printf 'banana\napple\npear\n' > fruit.txt
cat fruit.txt
```

`printf` interprets each `\n` in the format text as a new line.

`cat` displays the file's contents. Before using `>`, check `pwd` and the destination filename because replacing a file happens immediately, without confirmation requirements. To add text without replacing existing contents, use `>>`:

```bash
printf 'mango\n' >> fruit.txt
cat fruit.txt
```

For a quick view of a longer file, use `head`, `tail`, and `wc`:

```bash
head -n 2 fruit.txt
tail -n 1 fruit.txt
wc -l fruit.txt
```

`head -n 2` displays the first two lines, `tail -n 1` displays the last line, and `wc -l` counts lines. Each command reads `fruit.txt`; none of them changes it.

## Inspect distribution information

`cat` can also display a system information file. To identify the Linux distribution and version of the environment where this terminal is running, run:

```bash
cat /etc/os-release
```

`/etc/os-release` is a text file that identifies the distribution and its release. `cat` displays it without changing it. A Duckietown Workspace might report output such as:

```shell
PRETTY_NAME="Ubuntu 24.04.4 LTS"
NAME="Ubuntu"
VERSION_ID="24.04"
VERSION="24.04.4 LTS (Noble Numbat)"
VERSION_CODENAME=noble
ID=ubuntu
ID_LIKE=debian
```

`PRETTY_NAME` is a human-readable release name, `NAME` identifies the distribution, and `VERSION` identifies the release. The results apply only to this terminal's current environment; a base station, browser editor, service environment, and Duckiedrone can differ. [Notebook 2](./2-linux-shell-and-navigation.ipynb) introduces the corresponding `uname -m` architecture check.

## Copy, move, and remove files

Use `cp` to create a copy. Then use `mv` to give that copy a new name:

```bash
cp fruit.txt fruit_copy.txt
mv fruit_copy.txt fruit_backup.txt
ls -l
```

After `cp`, both filenames refer to separate files. After `mv`, `fruit_copy.txt` no longer exists and the copied file is named `fruit_backup.txt`. `ls -l` helps you check the exact filenames before removing anything.

Remove only the explicitly named practice copy:

```bash
rm fruit_backup.txt
ls -l
```

`rm` removes the named directory entry immediately and does not use a graphical Trash; recovery is not guaranteed. Check `pwd` and the exact filename first before proceeding. This LX never requires `rm -r`, which can remove a directory and everything inside it. Avoid broad commands such as `rm *`: the shell expands `*` to matching filenames before `rm` runs, so it can remove important files in the wrong directory.
