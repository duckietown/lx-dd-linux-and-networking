# Filenames, Wildcards, and Search

Clear filenames and focused searches make shell work safer. This notebook introduces filename conventions, wildcards, and focused search commands.

## Use clear filenames

Linux filenames are case-sensitive: `Fruit.txt` and `fruit.txt` are different names. Prefer lowercase names with underscores or hyphens, such as `sensor_log.txt` or `sensor-log.txt`, so the shell can handle them without extra quoting.

```bash
PRACTICE_DIR=/tmp/linux-search-practice
mkdir -p "$PRACTICE_DIR"
cd "$PRACTICE_DIR"
printf 'banana\napple\npear\n' > fruit.txt
printf 'first report\n' > report_1.txt
printf 'second report\n' > report_2.txt
```

## Match names with wildcards

Use wildcards to select the two report files:

```bash
cat report_?.txt
ls report_*.txt
```

`?` matches one character, so `report_?.txt` matches both report files. `*` matches zero or more characters, so `report_*.txt` matches names beginning with `report_` and ending with `.txt`. The shell expands a wildcard before it starts `cat` or `ls`.

Spaces make filenames harder to work with because the shell normally uses spaces to separate arguments. Quotation marks tell the shell to treat the enclosed text as one name:

```bash
printf 'draft\n' > "draft notes.txt"
cat "draft notes.txt"
```

The quoted filename works, but `draft_notes.txt` is usually easier to type and use in scripts.

## Find files and search text

Use `find` to search directory names and `grep` to search inside text files. Start `find` at `.` to search from the current directory:

```bash
find . -maxdepth 2 -type f -print
find . -name 'report_*.txt' -print
grep -n 'apple' fruit.txt
grep -n 'report' report_*.txt
```

`-maxdepth 2` keeps the result short, `-type f` selects regular files, and `-print` writes each matching path. The quotes around `'report_*.txt'` prevent the shell from expanding the wildcard before `find` receives it. `grep -n` prints matching lines with their line numbers.
