## Standard streams

The shell gives every command three standard streams:

- __Standard input__ (`stdin`) carries text into a program.

- __Standard output__ (`stdout`) carries a program's normal result.

- __Standard error__ (`stderr`) carries warnings and errors separately from the normal result.

Keeping normal output separate from errors lets a program's data remain usable while diagnostic messages stay visible or are recorded independently.

## Redirect input and output

Prepare a small file, then use redirection operators:

```bash
PRACTICE_DIR=/tmp/linux-stream-practice
mkdir -p "$PRACTICE_DIR"
cd "$PRACTICE_DIR"
printf 'banana\napple\npear\n' > fruit.txt
printf 'mango\n' >> fruit.txt
cat < fruit.txt
```

`>` redirects standard output and replaces a file's contents. `>>` appends standard output. `<` supplies a file as standard input. The `cat < fruit.txt` form displays the same data that `cat fruit.txt` would read as an argument, while making the input stream explicit.

## Separate normal output from errors

`2>` redirects standard error. The following intentionally fails only to demonstrate separation:

```bash
ls does-not-exist > listing.txt 2> errors.txt
cat listing.txt
cat errors.txt
```

The failed `ls` command writes no normal output to `listing.txt`; its error appears in `errors.txt`. This distinction is useful when a program's data must remain separate from a progress or diagnostic message.
