# Users, Ownership, and Permissions

Linux decides whether a process may read, change, or run a file by checking the current user, the file owner and group, and its permission bits. This notebook introduces users, ownership, and permissions.

## Identify the current account

```bash
whoami
id
```

`whoami` prints your current account name. `id` also shows its numeric user identifier (ID) and the groups it belongs to. In an isolated runtime environment, these values can differ from the account on the base station or Duckiedrone.

## Inspect and narrow permissions

Create a practice file and inspect its long listing before and after narrowing access for the group and all other users:

```bash
PRACTICE_DIR=/tmp/linux-permissions-practice
mkdir -p "$PRACTICE_DIR"
cd "$PRACTICE_DIR"
printf 'private practice note\n' > private_notes.txt
ls -l private_notes.txt
chmod go-rwx private_notes.txt
ls -l private_notes.txt
```

The first character of `ls -l` identifies the file type. The next nine characters are three groups of permissions: user (the owner), group, and other users. Within each group, `r` means read, `w` means write, and `x` means execute. `chmod go-rwx` removes read, write, and execute permission from group (`g`) and other users (`o`), leaving the owner unchanged. The final two groups should display `---`.

For a regular file, `x` allows it to run as a program or script. For a directory, `x` allows a user to enter it and access names inside it. Apply the smallest permission change that solves the problem. Do not respond to a permission error with `chmod 777` or `sudo`; inspect the owner, group, and intended access first.
