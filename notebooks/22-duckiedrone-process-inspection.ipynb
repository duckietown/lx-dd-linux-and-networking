# Duckiedrone Process Inspection

This notebook introduces read-only process inspection on a Duckiedrone.

## Inspect Duckiedrone services

A physical Duckiedrone and a virtual Duckiedrone each have their own process view. A process list after connecting shows the processes visible in that Duckiedrone environment rather than every process on the base station.

Commands run at one of three levels:

```text
base-station shell -> shell on the Duckiedrone -> service-specific shell
```

Run `dts` on the base station. Service-specific commands run in a relevant service environment or workbench where that service's tools are available.

From one authorized physical or virtual Duckiedrone shell opened in Notebook 19 or Notebook 20, run:

```bash
ps -ef
```

`ps -ef` lists processes visible to the current shell. A standard Duckiedrone provides three broad groups of services:

- Core platform services that store settings and present device status.

- Device drivers and services such as `dtps`, `driver-camera`, `driver-tof-bottom`, and `dashboard`.

- Transport and bridge services that connect device data to other platform components.

Additional sensor, altitude, and visualization services run only when selected during setup, so their absence alone does not prove a failure. The Dashboard at `http://DUCKIEDRONE_NAME.local` shows Duckiedrone status. Use it and process observations to investigate a problem before escalating it through the authorized support path.

## Monitor processes read-only

First check whether `htop` is available in the authorized Duckiedrone shell:

```bash
command -v htop
```

If it prints a path such as `/usr/bin/htop`, start the monitor in read-only mode:

```bash
htop --readonly
```

`htop --readonly` refreshes processes and CPU, memory, and swap use while disabling process-changing features. It shows processes visible to the current shell: an SSH shell or a virtual Duckiedrone shell shows the Duckiedrone environment, while a service-specific shell shows only its service environment. Treat high CPU or memory use as evidence to investigate, not a diagnosis. Press `q` or `F10` to leave.

A physical Duckiedrone might not include `htop`. If `command -v htop` prints nothing, do not install it for this lesson. Use the standard read-only fallback:

```bash
top
```

Press `q` to leave `top`; do not use its process-control keys. Record observations or inspect Dashboard and service status before following an authorized troubleshooting procedure.
