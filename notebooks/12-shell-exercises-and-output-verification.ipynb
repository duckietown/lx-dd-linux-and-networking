## Complete the stream exercises

Each remaining file deliberately raises `NotImplementedError` until you replace the body of `main()`. Complete these exercises in order:

1. Implement `my_cat.py` so it copies standard input to standard output unchanged.

2. Implement `my_reverse_sort.py` so it reads input lines, sorts them in reverse alphabetical order, and writes the result to standard output.

3. Implement `my_sort_status.py` so it writes sorted input to standard output and a brief status message to standard error.

Test `my_cat.py` with terminal input and a pipe:

```bash
python3 packages/shell_exercises/my_cat.py
printf 'one\ntwo\n' | python3 packages/shell_exercises/my_cat.py
```

After typing input for the first command, press `Ctrl-D` on an empty prompt to send an __end-of-file (EOF)__ signal. It tells the program that no more input is coming, so the program can process what you typed. `Ctrl-C` instead stops the program immediately.

Test reverse sorting with known input:

```bash
printf 'pear\napple\nbanana\n' | python3 packages/shell_exercises/my_reverse_sort.py
```

The result should list `pear`, then `banana`, then `apple`. Finally, verify that `my_sort_status.py` keeps sorted output separate from its status message:

```bash
printf 'pear\napple\nbanana\n' | python3 packages/shell_exercises/my_sort_status.py > "$PRACTICE_DIR/sorted.txt" 2> "$PRACTICE_DIR/status.txt"
cat "$PRACTICE_DIR/sorted.txt"
cat "$PRACTICE_DIR/status.txt"
```

`sorted.txt` should contain only sorted lines, while `status.txt` should contain only the program's status message.

## Compare expected and actual output

Move to the temporary directory, create the expected reverse-sorted output, run your implementation from the saved LX root, and compare the two files:

```bash
cd "$PRACTICE_DIR"
printf 'pear\nbanana\napple\n' > expected_reverse.txt
printf 'pear\napple\nbanana\n' | python3 "$LX_ROOT"/packages/shell_exercises/my_reverse_sort.py > actual_reverse.txt
diff -u expected_reverse.txt actual_reverse.txt
```

`diff` compares two files. `-u` uses a unified format that labels expected lines with `-` and actual lines with `+`. When the files match, `diff -u` prints nothing and returns exit status `0`. When they differ, it prints differences and returns `1`; here that is useful evidence about what to fix, not a problem with `diff`.
