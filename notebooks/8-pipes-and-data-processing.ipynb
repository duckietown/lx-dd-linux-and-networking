# Pipes and Data Processing

A pipe composes small programs into a data-processing sequence. The command on the left writes standard output (`stdout`), and the command on the right reads that text as standard input (`stdin`).

## Send text to another command

```bash
printf 'pear\napple\nbanana\n' | sort
printf 'pear\napple\nbanana\n' | sort -r
```

The programs do not need to know about each other. They agree only that the stream contains text lines. The first command sorts alphabetically; `-r` reverses the order.

## Compose a short processing pipeline

Create a self-contained input file, select matching lines, then sort only those lines:

```bash
PRACTICE_DIR=/tmp/linux-pipe-practice
mkdir -p "$PRACTICE_DIR"
cd "$PRACTICE_DIR"
printf 'banana\napple\npear\nmango\n' > fruit.txt
cat fruit.txt | grep 'e' | sort
```

Read the pipeline from left to right: `cat` writes each fruit line, `grep 'e'` keeps lines containing `e`, and `sort` orders the remaining lines. Keep each stage focused so its input and output are easy to inspect.
