# Pipes and Data Processing

A __pipe__ composes small programs into a data-processing sequence. The command on the left writes standard output (`stdout`), and the command on the right reads that text as standard input (`stdin`).

## Send text to another command

Send unsorted text to `sort` and reverse-sort it:

```bash
printf 'pear\napple\nbanana\n' | sort
printf 'pear\napple\nbanana\n' | sort -r
```

The programs do not need to know about each other. They agree only that the stream contains text lines. The first command sorts alphabetically; `-r` reverses the order.

## Compose a short processing pipeline

Create a self-contained input file, select matching lines, then sort only those lines:

```bash
PRACTICE_DIR=/tmp/linux-pipe-practice
mkdir -p "$PRACTICE_DIR"
cd "$PRACTICE_DIR"
printf 'banana\napple\npear\nmango\n' > fruit.txt
grep 'e' fruit.txt | sort
```

Read the pipeline from left to right: `grep 'e' fruit.txt` reads the file and keeps lines containing `e`, then `sort` orders the remaining lines. Keep each stage focused so its input and output are easy to inspect.

## Check pipeline outcomes

In Bash, a pipeline normally reports the exit status of its last command. A later stage can therefore report success even when an earlier stage failed to read or produce the intended data. When every stage of a trusted Bash script must succeed, use `set -o pipefail` near the start of that script. With `pipefail` enabled, a pipeline reports a nonzero status when any stage fails.

Some filters use a nonzero status for an expected condition. For example, `grep` returns `1` when it finds no matching lines. Decide whether that result should stop the script before relying on `pipefail`.
