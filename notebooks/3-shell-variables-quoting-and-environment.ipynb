# Shell Variables, Quoting, and Environment

Variables enable reusing values, customizing commands, and controlling how programs behave. Understanding quoting and environment variables helps write reliable commands and avoid unexpected results. 

_Shell variables_ name pieces of text for the current terminal. _Environment variables_ are copied into programs the shell starts. This notebook explains variables, quoting, and the environment.

## Work with shell variables

Let's navigate back to our test environment:

```bash
cd /tmp/linux-shell-practice
```

and set a label for this terminal, then display it with the home and current-directory values:

```bash
PRACTICE_LABEL='Linux shell practice'
printf 'Label: %s\n' "$PRACTICE_LABEL"
printf 'Home directory: %s\n' "$HOME"
printf 'Current directory: %s\n' "$PWD"
```

`PRACTICE_LABEL` is the variable name. The `=` saves the literal text `Linux shell practice` as its value, so do not put spaces around it. `$PRACTICE_LABEL` (note the `$` before the variable name) tells the shell to use the saved value. `$HOME` names your home directory and `$PWD` names the current directory. A variable such as `PRACTICE_LABEL` belongs to this shell and lasts only while this terminal remains open, so if you open a new terminal and try:

```bash
echo $PRACTICE_LABEL
```

you will get a blank line (empty string).

## Quote values deliberately

Compare the output from a literal string with the saved label:

```bash
printf 'Literal text: $PRACTICE_LABEL\n'
printf "Expanded value: $PRACTICE_LABEL\n"
```

The first result includes the characters `$PRACTICE_LABEL`; the second includes the saved value, `Linux shell practice`. Single quotes keep `$PRACTICE_LABEL` literal, while double quotes expand it and keep the resulting value as one argument. Double-quote a path variable such as `"$PRACTICE_DIR"` so a path containing spaces remains one argument; the quotes are not part of the path.

## Find commands and export values

`PATH` is an system predefined environment variable listing the directories the shell searches when you type a command without a path. Use `command -v` to see what the shell would run:

```bash
command -v python3
command -v cd
```

`python3` is an executable file, while `cd` is built into the shell. You normally do not need to change `PATH` in this learning experience.

A shell copies an environment variable into each command it starts. Use `export` only when a child program needs a value; it does not make the value permanent:

```bash
PRACTICE_MODE=practice
printf 'Current shell: %s\n' "$PRACTICE_MODE"
env | grep '^PRACTICE_MODE=' || printf '%s\n' 'Not exported yet'
export PRACTICE_MODE
env | grep '^PRACTICE_MODE='
```

`env` shows the environment received by a command. Before `export`, `grep` finds no `PRACTICE_MODE` entry; after it, the entry appears as `PRACTICE_MODE=practice`. A child program receives a copy of an exported value, so it cannot change the current shell's variables. Closing the terminal removes both forms.
