# Linux Shell and Navigation

Linux is the operating system behind the Duckiedrone and many development environments. This notebook introduces the terminal, the shell, and the filesystem locations that make commands predictable.

## Know where commands run

A terminal is the window in which you type. A shell is the program inside that terminal that reads your commands. The shell has a current working directory, the default location for file operations. Start at the LX root, the directory that contains `packages/`, and inspect it before changing files:

```bash
pwd
ls
```

`pwd` prints the path of the current working directory. `ls` lists the names of files and directories in that directory. Do not continue until `ls` shows `packages`. This notebook does not need `sudo`, a command that runs another command with administrator permissions. Do not add `sudo` unless the instructions explicitly ask for it.

## Create a safe practice area

Use a temporary directory for practice rather than the LX files you will edit later. Create it and work there:

```bash
mkdir -p /tmp/linux-shell-practice
cd /tmp/linux-shell-practice
pwd
```

`mkdir -p` attempts to create the requested directory and any missing parent directories. It does not report an error if the requested directory already exists. `cd /tmp/linux-shell-practice` changes the current working directory to that path, so the following `pwd` prints it.

## Navigate with paths

A path names a file or directory. A path beginning with `/` is absolute: it starts at the filesystem root and means the same thing from every directory. A path without a leading `/` is relative: it starts from the current working directory. Create a small tree and move through it:

```bash
mkdir -p notes/drafts
ls
cd notes
pwd
cd drafts
pwd
cd ..
pwd
cd ../..
pwd
```

`.` means the current directory and `..` means its parent. From `/tmp/linux-shell-practice/notes/drafts`, `cd ..` moves to `notes`. In the command sequence above, `cd ../..` then starts in `/tmp/linux-shell-practice/notes` and ends in `/tmp`.

Your home directory has the shortcut `~`; `cd` with no argument changes the current working directory to your home directory. Return to the practice area with `cd /tmp/linux-shell-practice`. Use `ls -la` when you need details: `-l` shows ownership, permissions, and file size, while `-a` includes names that begin with `.`, which Linux normally hides.

## Mount points

A __mount point__ is a directory where Linux makes another filesystem or directory tree available. It remains part of the pathname hierarchy, but the files beneath it can be stored or managed elsewhere. Later examples use this idea to make a selected directory available in another isolated environment.
