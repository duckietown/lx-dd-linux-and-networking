# Linux Shell and Navigation

Linux is the operating system behind many development environments, including the Duckiedrone. Its kernel manages hardware and provides system services to programs. A shell is a user-space program that lets users issue commands and interact with those services.

This notebook introduces the terminal, the shell, and the filesystem locations that make commands possible and enable users to predictably interact with computers.

## A technical foreword - do not use `sudo`

This notebook should never require using `sudo`, a command that runs another command with administrator permissions. Do not add `sudo` unless the instructions explicitly ask for it.

## Know where commands run

Modern users often interact with computers by tapping apps or clicking folders. Older generations may remember when interacting with computers largely meant typing instructions, for example in DOS (the "Disk Operating System").

In robotics, most actions are "old school", as in performed by typing commands in a window called _terminal_. It turns out that mastering terminal-fu is the most efficient of ways to interact with machines (so far?). All computer operating systems have some kind of terminal program. In Ubuntu and macOS, the out of the box programs are called Terminal, in Windows instead Powershell. Focusing on Linux:

* A _shell_ is the program inside that terminal that reads your commands. Example shells are GNU Bourne-Again Shell (`bash`), the default shell in most Linux environments, and Z Shell (`zsh`).

* The shell has a current _working directory_: the default location for file operations. 

Open a Terminal (in a [Duckietown Workspace](https://docs.duckietown.com/ente/duckietown-manual/10-setup/00-computer/setup-duckietown-workspace.html), in the example below), and you will be met by:

```bash
ubuntu@orbstack:~$ 
```

The exact format will depend on the shell used, but critical elements are displayed:

* The user, with username: `ubuntu` 
* The hostname: `orbstack`
* The working directory: `~` (in Linux, `~` is the shortcut for `/home/username`, in this case `/home/ubuntu`). Note that the root directory in a Linux system is indicated by `/`. 

Let us go exploring.

Where your cursor is blinking (sometimes it doesn't), i.e., after the `$` sign, type `pwd`. At the beginning we will show these kind of passages explicitly, e.g.:

```bash
ubuntu@orbstack:~$ pwd
```

Moving forward, we will assume the path part, and for simplicity just write:

```bash
pwd
```

This first command, `pwd` prints the **p**ath of the current **w**orking **d**irectory. If your working directory is `~`, you will get `/home/ubuntu` - as predicted above. To list what is contained in the directory:

```bash
ubuntu@orbstack:~$ ls
```

`ls` lists the names of visible files and folders in that directory. To see all contents, including those files and directories marked as invisible (to, e.g., avoid accidental erasing), list **a**ll instead:

```bash
ubuntu@orbstack:~$ ls -a
```

If there are hidden files or folders, you will see their names listed, preceded by a `.`. For example, `.bashrc` is a configuration file for the `bash` shell, that gets executed every time a terminal is opened. But let us not digress. 


Now navigate to the root folder of this learning experience, i.e., the same folder where you typed `dts code editor` to open this browser session, by calling the **c**orrect **d**irectory path. It should be something like this (but `~/path-to-duckietown-lxs/` will depend on where you cloned the GitHub repository of this LX):

```bash
ubuntu@orbstack:~$ cd ~/path-to-duckietown-lxs/lx-dd-linux-and-networking
```

## Navigate with paths

A path names a file or directory. A path beginning with `/` is absolute: it starts at the filesystem root and means the same thing from every directory. If you notice above, we navigated in this LX directory, but the `/linux-shell-practice` one was generate outside of it.

A path without a leading `/` is relative: it starts from the current working directory. For example, with the commands below we create a small tree and move through it:

```bash
ls 
mkdir -p notes/drafts
ls 
cd notes
pwd
cd drafts
pwd
cd ..
pwd
cd ../..
pwd
```

The full stop (`.`) means "the current directory", and `..` means its parent (one above / back). 

From `/tmp/linux-shell-practice/notes/drafts`, `cd ..` moves to `notes`. In the command sequence above, `cd ../..` then starts in `/tmp/linux-shell-practice/notes` and ends in `/tmp`.

Your home directory has the shortcut `~`; `cd` with no argument changes the current working directory to your home directory. 

Return to the practice area with `cd /tmp/linux-shell-practice`. Use `ls -la` when you need details: `-l` shows ownership, permissions, and file size, while `-a` includes names that begin with `.`, which, as previously discussed, Linux normally hides.

```text
ubuntu@orbstack:/tmp/linux-shell-practice$ ls -la
total 0
drwxr-xr-x 1 ubuntu ubuntu   10 Sep 20 09:03 .
drwxrwxrwt 1 root   root   2148 Sep 21 07:32 ..
drwxr-xr-x 1 ubuntu ubuntu   12 Sep 20 09:03 notes
```

In this case `ls -la` is showing the current and parent folders (`.`, `..`), and the `notes` directory. 

```text
drwxr-xr-x 1 ubuntu ubuntu 10 Sep 20 09:03 notes
│─────────│ │ │────│ │────│ │  │───────────│ │───│
permissions │ owner  group  size modified time name
            │
            link count
```

* drwxr-xr-x: file type and permissions. The first character is the type: `d` means directory; `-` would mean a regular file. 
* The remaining characters are permissions for the owner, group, and everyone else.
* `1`: number of [hard links](https://en.wikipedia.org/wiki/Hard_link) to the item.
* `ubuntu`: user who owns it.
* `ubuntu`: group that owns it.
* `10`: size in bytes. For a directory, this is metadata size—not the total size of its contents.
* `Sep 20 09:03` — when it was last modified.
* `notes`: its name.

## Mount points

A __mount point__ is a directory where Linux makes another filesystem or directory tree available. It remains part of the pathname hierarchy, but the files beneath it can be stored or managed elsewhere. Later examples use this idea to make a selected directory available in another isolated environment.
