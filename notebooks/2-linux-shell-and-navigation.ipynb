# Linux Shell and Navigation

Linux is the operating system behind many development environments, including the Duckiedrone. Its kernel manages hardware and provides system services to programs. A shell is a user-space program that lets users issue commands and interact with those services.

This notebook introduces the terminal, the shell, and the filesystem locations that make commands possible and enable users to predictably interact with computers.

## A technical foreword - do not use `sudo`

This notebook never requires using `sudo`, a command that runs another command with administrator permissions. Do not add `sudo` to its commands.

## Know where commands run

Modern users often interact with computers by tapping apps or clicking folders. Older generations may remember when interacting with computers largely meant typing instructions, for example in Disk Operating System (DOS).

In robotics, many development and operational tasks are performed by entering commands in a window called a __terminal__. Familiarity with the command line makes these tasks more efficient. Many operating systems provide a terminal application or command-line environment. The GNU's Not Unix (GNU) project develops Bash, a common Linux shell. Ubuntu Desktop 24.04 includes a terminal emulator from the GNU Network Object Model Environment (GNOME) desktop environment: GNOME Terminal. Other Ubuntu installations may use a different terminal emulator. macOS includes Terminal. Windows Terminal can run shells such as PowerShell. On Linux:

- A __shell__ is the program inside that terminal that reads your commands. Example shells include Bash (`bash`), the default shell in most Linux environments, and Z Shell (`zsh`).

- The shell has a current __working directory__: the default location for file operations. 

Open a terminal. It should show the following command prompt if using a [Duckietown Workspace](https://github.com/duckietown/workspace), or similar if not:

```shell
ubuntu@orbstack:~$ 
```

The exact format will depend on the shell used, but critical elements are displayed:

- The user, with username: `ubuntu` 

- The hostname: `orbstack`

- The working directory: `~` (`~` is the shell shortcut for the current user’s home directory, `/home/ubuntu` in this example). The filesystem root is indicated by the leftmost `/`. 

Moving forward, the command prompt will only be shown in the relevant context.

To run a command, select the terminal (your cursor should be visible after the `$` sign), type the command, and press `Enter`.

To print the path of the current working directory, run:

```bash
pwd
```

If your working directory is `~`, you will get the following, as predicted above:

```shell
/home/ubuntu
```

To list the names of visible files and folders in that directory, run:

```bash
ls
```

Names beginning with `.` are hidden by convention; use `ls -a` to list all entries, including them:

```bash
ls -a
```

If there are hidden files or folders, you will see their names listed, preceded by a `.`. For example, `.bashrc` is a configuration file that interactive Bash shells commonly read when they start. But let us not digress. 

To report the machine architecture used by the running kernel, run:

```bash
uname -m
```

For example, a 64-bit ARM Duckietown Workspace might report:

```shell
aarch64
```

`aarch64` is the name `uname` uses for 64-bit ARM, which Linux distributions commonly label `arm64`. The result applies only to this terminal's current environment; a base station, browser editor, service environment, and Duckiedrone can differ.

Now navigate to the root folder of this LX using `cd` (change directory), where `/path/to` is the path to the parent folder of the LX:

```bash
cd /path/to/lx-dd-linux-and-networking
```

Do not continue until `pwd` ends with `/lx-dd-linux-and-networking`.

## Navigate with paths

A __path__ names a file or directory. A path beginning with `/` is absolute: it starts at the filesystem root and means the same thing from every directory. Above, we navigated within the LX directory, but `/tmp/linux-shell-practice` was created outside it.

A path without a leading `/` is relative: it starts from the current working directory. For example, with the commands below we create a small tree and move through it:

```bash
ls 
mkdir -p notes/drafts
ls 
cd notes
pwd
cd drafts
pwd
cd ..
pwd
cd ../..
pwd
```

The full stop (`.`) means the current directory, and `..` means its parent directory.

From `/tmp/linux-shell-practice/notes/drafts`, `cd ..` moves to `notes`. In the command sequence above, `cd ../..` then starts in `/tmp/linux-shell-practice/notes` and ends in `/tmp`.

Your home directory has the shortcut `~`; `cd` with no argument changes the current working directory to your home directory. 

Return to the practice area with `cd /tmp/linux-shell-practice`. Use `ls -la` when you need details: `-l` shows ownership, permissions, and file size, while `-a` includes names that begin with `.`, which, as previously discussed, Linux normally hides:

```shell
ubuntu@orbstack:/tmp/linux-shell-practice$ ls -la
total 0
drwxr-xr-x 1 ubuntu ubuntu   10 Sep 20 09:03 .
drwxrwxrwt 1 root   root   2148 Sep 21 07:32 ..
drwxr-xr-x 1 ubuntu ubuntu   12 Sep 20 09:03 notes
```

In this case `ls -la` is showing the current and parent folders (`.`, `..`), as well as the `notes` directory.

From left to right, the fields in this listing are:

- `drwxr-xr-x`: file type and permissions. The first character is the type: `d` means directory; `-` would mean a regular file. 

- `1`: number of [hard links](https://en.wikipedia.org/wiki/Hard_link) to the item.

- `ubuntu`: user who owns it.

- `ubuntu`: group that owns it.

- `10`: size in bytes. For a directory, this is metadata size—not the total size of its contents.

- `Sep 20 09:03`: when it was last modified.

- `notes`: its name.

## Mount points

A __mount point__ is a directory where Linux makes another filesystem or directory tree available. It remains part of the pathname hierarchy, but the files beneath it can be stored or managed elsewhere. Later examples use this idea to make a selected directory available in another isolated environment.
