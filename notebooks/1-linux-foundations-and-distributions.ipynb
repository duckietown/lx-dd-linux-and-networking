## Kernel and distribution definition

 The word "Linux" commonly refers to the Linux _kernel_: the core software that manages hardware and provides system services to programs. A complete usable operating system combines that kernel with user-space tools, libraries, services, package management, and applications.

## A layered operating system

A useful simplified model is:

```text
hardware
  |
  v
Linux kernel: central processing unit (CPU) scheduling, memory, devices, networking, filesystems
  |
  v
user space: system services, libraries, package tools
  |
  v
shells, desktop applications, and robot software
```

The kernel runs with privileges to control hardware. Most programs, including the _shell_ you will use next, run in user space and request kernel services through defined interfaces. The exact tools and defaults above the kernel vary by system.

## Kernel and processor (CPU) architecture

The kernel is software; CPU architecture describes the _instruction set of the hardware_ that runs it. Common examples include `x86_64` (also called `x64` or `AMD64`, and used by many Windows development computers) and `arm64`, used by many embedded computers. A Linux distribution can support more than one CPU architecture, while one computer runs a build chosen for its own architecture. The [Linux Kernel documentation](https://docs.kernel.org/) maintains separate architecture documentation for the ARM processor architecture family, its 64-bit ARM64 variant, x86, and other platforms.

This distinction matters for Duckiedrone work: a base-station shell, an isolated service environment, and a Duckiedrone shell can refer to different filesystems, processes, and networks. The base station and service environments can also use a different distribution or CPU architecture. The Duckiedrones covered by this LX use a known operating-system platform and CPU architecture, so do not treat the terminal prompt alone as proof that two non-Duckiedrone environments are the same.

## Inspect the current environment

Before following an installation guide, open a terminal (on your Linux machine, or in the `dts code editor` VS code browser-based environment, or (best) in your [Duckietown Workspace](https://docs.duckietown.com/ente/duckietown-manual/10-setup/00-computer/setup-duckietown-workspace.html)) and inspect the environment that runs the command:

```bash
uname -m
cat /etc/os-release
```

`uname -m` reports the machine architecture used by the running kernel. `/etc/os-release` identifies the distribution and version. These results describe only the current shell context. Use these checks to distinguish base-station, editor, and service environments when investigating compatibility issues; the Duckiedrone platform itself is already known.

For example, opening a terminal in a [Duckietown Workspace](https://docs.duckietown.com/ente/duckietown-manual/10-setup/00-computer/setup-duckietown-workspace.html) yields:

```text
ubuntu@orbstack:~$ uname -m
aarch64 

ubuntu@orbstack:~$ cat /etc/os-release 
PRETTY_NAME="Ubuntu 24.04.4 LTS"
NAME="Ubuntu"
VERSION_ID="24.04"
VERSION="24.04.4 LTS (Noble Numbat)"
VERSION_CODENAME=noble
ID=ubuntu
ID_LIKE=debian
HOME_URL="https://www.ubuntu.com/"
SUPPORT_URL="https://help.ubuntu.com/"
BUG_REPORT_URL="https://bugs.launchpad.net/ubuntu/"
PRIVACY_POLICY_URL="https://www.ubuntu.com/legal/terms-and-policies/privacy-policy"
UBUNTU_CODENAME=noble
LOGO=ubuntu-logo
```

Where `aarch64` is the architecture (it stands for `arm64`), `Ubuntu` is the Linux distribution, and `24.04.4 LTS (Noble Numbat)` the version used. 

## What a distribution adds

A Linux distribution, or distro, packages the kernel with a selected set of user-space software, software repositories, package-management tools, defaults, release policy, and maintenance process. [Debian](https://www.debian.org/), [Ubuntu](https://ubuntu.com/), and [Fedora](https://fedoraproject.org/) are examples of distributions. They can use the same Linux kernel family while differing in package formats, package commands, available versions, and configuration conventions.

Therefore, first identify the distribution and version before following an installation guide. A command documented for one distribution is not automatically appropriate for another. This LX begins with portable inspection and shell concepts; later notebooks explain safe package, process, filesystem, and network work in the environment you actually control.
