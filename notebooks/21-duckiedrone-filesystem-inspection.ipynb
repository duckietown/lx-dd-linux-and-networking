# Duckiedrone Filesystem Inspection

This notebook introduces filesystem inspection on a Duckiedrone.

## Inspect filesystem landmarks

After opening one authorized physical or virtual Duckiedrone shell in Notebook 19 or Notebook 20, use that shell for the following read-only inspection.

Begin at `/`, the root directory of that Duckiedrone filesystem. These commands display names and metadata only:

```bash
cd /
pwd
hostname
whoami
ls -ld /home /home/duckie /code /data /etc /var /tmp /proc /sys /dev
ls -la /home/duckie
ls -la /code
ls -la /data
```

`/` is the filesystem root. It is different from `/root`, the home directory for the `root` account. These landmarks are useful:

- `/home/duckie` is the normal home directory for the `duckie` account.

- `/code` is where Duckietown developer tooling can synchronize source code. It can be empty or hold only a marker until code is synchronized.

- `/data` contains Duckiedrone configuration, logs, recordings, stacks, and other data used by the device and its services.

- `/etc` holds system configuration, `/var` holds changing service data and logs, and `/tmp` is for temporary files.

- `/proc`, `/sys`, and `/dev` expose information and interfaces from the Linux kernel; they are not ordinary places to store work.

## Notice the user context

A physical Duckiedrone SSH session normally uses the `duckie` account. A shell from `dts duckiebot virtual connect` currently starts as `root`, so `whoami` can differ. This administrative detail of the virtual device is not a reason to use `root` for normal work on a physical Duckiedrone.

Do not inspect `/secrets`, `/root`, or the contents of `/data/config`; they can hold sensitive or device-specific information. Keep observation limited to authorized paths and metadata.

## Further reading

The [Filesystem Hierarchy Standard](https://refspecs.linuxfoundation.org/FHS_3.0/fhs/index.html) describes conventional Linux filesystem locations.
