# Virtual Duckiedrone Connections

SSH lets you work on a physical Duckiedrone, but practicing Linux commands in Duckietown does not require a physical Duckiedrone powered on beside you.

A __virtual Duckiedrone__ provides a local Duckiedrone software environment on your base station. You can open its shell and practice inspecting files and processes much as you would on a physical Duckiedrone.

In this notebook, you will create (or select, if already created) a virtual Duckiedrone, start it, open its shell, and stop it when finished.

Run all `dts` commands below in the __base-station terminal__, after completing the prerequisites in the [LX setup instructions](../README.md#run-this-lx).

## Create a virtual Duckiedrone

First, check whether you already have a virtual Duckiedrone available:

```bash
dts duckiebot virtual list
```

If you already have a suitable virtual Duckiedrone for this activity, use its name and skip creation.

Otherwise, choose a [valid name](https://docs.duckietown.com/ente/opmanual-dd24/30-duckiedrone-setup-and-config/02-software-initialization-advanced.html#duckiedrone-hostname-naming-constraints) containing only lowercase Latin letters and digits (no special characters) that distinguishes it from your physical Duckiedrone:

```bash
dts duckiebot virtual create -t duckiedrone -c DD24 DUCKIEDRONE_NAME
```

Although the command begins with `duckiebot`, `-t duckiedrone` selects the Duckiedrone type. `-c DD24` selects its compatible platform configuration and software stack.

The command asks you to enter and confirm a password for the virtual Duckiedrone's `duckie` account. Choose a personal password with at least eight characters. Enter it at the prompts; typed characters are not displayed.

The creation process might take a few minutes to prepare the software environment. Once created, starting the virtual Duckiedrone, connecting to it, and doing [other supported operations](https://docs.duckietown.com/ente/duckietown-manual/50-duckiematrix/virtual-duckietown-robots/introduction-to-virtual-duckiebots.html) are separate steps. [Figure 1](#figure-1) shows what happens when you open and exit its shell.

## List, start, and connect

Once you have created or selected a virtual Duckiedrone, inspect its current status:

```bash
dts duckiebot virtual list
```

Your virtual Duckiedrone should appear by its chosen name. For illustration, the virtual Duckiedrone `vamelia` appears alongside the virtual Duckiebot `vargo`:

```shell
         |    Type     | Model |    Status    | Uptime
-------- | ----------- | ----- | ------------ | ------
vamelia  | duckiedrone |  DD24 |     Down     |      -
vargo    |   duckiebot | DB21J |     Down     |      -
```

If your virtual Duckiedrone is not running (`Status`: `Down`), start it:

```bash
dts duckiebot virtual start DUCKIEDRONE_NAME
```

Then open its shell:

```bash
dts duckiebot virtual connect DUCKIEDRONE_NAME
```

This command checks that the virtual Duckiedrone is running and opens an interactive Bash shell inside it. It does not use SSH or request the `duckie` password entered during creation. The prompt should show your Duckiedrone's hostname. For `vamelia`, it would look like:

```shell
root@vamelia:/#
```

### Try it: check the virtual shell

Establish where you are:

```bash
hostname
whoami
pwd
```

Read these results as three separate pieces of information, as shown in [Table 1](#table-1).

It lists the current directory inside the virtual Duckiedrone, not your base station's project directory.

If `connect` reports that the Duckiedrone is not running, check its status and start the same named Duckiedrone. You do not need to create another one.

## Stop a virtual Duckiedrone

When you have finished practicing, stop your virtual Duckiedrone from the base-station terminal, using the same name as before:

```bash
dts duckiebot virtual stop DUCKIEDRONE_NAME
dts duckiebot virtual list
```

Confirm that the named Duckiedrone is no longer running.

Stopping preserves the virtual Duckiedrone for later use. Next time, start it and connect again; creation is not a step you repeat for every session.

The [Duckiematrix command reference](https://docs.duckietown.com/ente/devmanual-duckiematrix/intermediate/virtual_duckietown_robots/dts-commands.html) documents this lifecycle.

## Local virtual and remote connections

Both workflows give you a shell in a Duckiedrone environment, but they reach it differently, as [Table 2](#table-2) shows.

<table id="table-2" class="lx-table">
  <caption>Table 2: Physical and virtual Duckiedrone connection methods.</caption>
  <thead>
    <tr><th>Physical Duckiedrone</th><th>Local virtual Duckiedrone</th></tr>
  </thead>
  <tbody>
    <tr><td>Connect with <code>ssh duckie@DUCKIEDRONE_NAME.local</code></td><td>Connect with <code>dts duckiebot virtual connect DUCKIEDRONE_NAME</code></td></tr>
    <tr><td>Requires a network path to the Duckiedrone</td><td>Uses the local virtual-Duckiedrone environment</td></tr>
    <tr><td>Checks the server's identity and authenticates your account</td><td>Opens a shell through the local management command</td></tr>
    <tr><td><code>exit</code> closes the SSH session</td><td><code>exit</code> closes the virtual-Duckiedrone shell</td></tr>
  </tbody>
</table>

The virtual Duckiedrone runs locally but can still generate network traffic. Connections it makes to external services use the base station's networking infrastructure and remain subject to the network's access rules.
