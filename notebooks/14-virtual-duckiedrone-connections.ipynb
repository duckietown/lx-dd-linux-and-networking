# Virtual Duckiedrone Connections

A virtual Duckiedrone runs locally on the base station. This notebook introduces the Duckietown Shell (`dts`) virtual Duckiedrone workflow.

## Create a virtual Duckiedrone

To create a virtual Duckiedrone, run the following command, where `DUCKIEDRONE_NAME` is the name of the virtual Duckiedrone:

```bash
dts duckiebot virtual create -t duckiedrone -c DD24 DUCKIEDRONE_NAME
```

`-t duckiedrone` selects the Duckiedrone type, and `-c DD24` selects the DD24 configuration, which determines the virtual Duckiedrone's compatible platform profile and software stack.

The command prompts twice for a password for the virtual Duckiedrone's `duckie` account. Choose a personal password with at least eight characters and enter it only at those interactive prompts. Do not put it in a notebook, command, shell history, chat message, or source file.

This creates a virtual Duckiedrone DD24 on the base station. Start it before connecting. Unlike SSH, this workflow opens an interactive shell on a virtual Duckiedrone running locally on the base station; it does not create a network connection to a remote device.

## List, start, and connect

Inspect an existing virtual Duckiedrone's status:

```bash
dts duckiebot virtual list
```

If `DUCKIEDRONE_NAME` is not running, start it:

```bash
dts duckiebot virtual start DUCKIEDRONE_NAME
```

Then open a shell:

```bash
dts duckiebot virtual connect DUCKIEDRONE_NAME
```

`dts duckiebot virtual connect` verifies that the requested virtual Duckiedrone is running, then opens an interactive Bash shell on it. It does not prompt for the password chosen during creation; that password configures the virtual Duckiedrone's `duckie` account.

Use `exit` when finished to return to the base-station shell.

## Stop a virtual Duckiedrone

When you no longer need a running virtual Duckiedrone, stop only the named device from the base-station terminal:

```bash
dts duckiebot virtual stop DUCKIEDRONE_NAME
```

`dts duckiebot virtual stop` asks the named virtual Duckiedrone to shut down. Use `dts duckiebot virtual list` to confirm its status.

## Local virtual and remote connections

A physical Duckiedrone needs an authorized network route, identity verification, authentication, and SSH. A virtual Duckiedrone connection opens an interactive shell on a local virtual Duckiedrone, so `ssh-copy-id` is not part of this workflow. Network traffic produced by the virtual Duckiedrone still leaves through the base station and remains subject to the applicable network policy.
