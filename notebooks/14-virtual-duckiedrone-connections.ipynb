# Virtual Duckiedrone Connections

SSH lets you work on a physical Duckiedrone, but practicing Linux commands in Duckietown does not require having a physical robot powered on beside you.

A **virtual Duckiedrone** provides a local robot software environment on your base station. You can open its shell and practice inspecting files and processes before working on physical hardware, in pretty much the same way as you would with a physical Duckiedrone. 

In this notebook, you will create (or select, if already created) a virtual Duckiedrone, start it, open its shell, and stop it when finished.

Run all `dts` commands below in the **base-station terminal**. This assumes Docker and the Duckietown Shell are already configured, as described in the [LX setup instructions](../README.md#run-this-learning-experience).

## Create a virtual Duckiedrone

First, check whether you already have a virtual robot available:

```bash
dts duckiebot virtual list
```

If you already have a suitable virtual Duckiedrone for this activity, use its name and skip creation.

Otherwise, choose a [valid name](https://docs.duckietown.com/ente/opmanual-dd24/30-duckiedrone-setup-and-config/02-software-initialization-advanced.html#duckiedrone-hostname-naming-constraints) (only lower case latin letter and digits allowed, no special characters) that distinguishes it from your physical robot. We will use `vamelia` but you can replace it throughout this notebook if you choose another name:

```bash
dts duckiebot virtual create -t duckiedrone -c DD24 vamelia
```

Although the command begins with `duckiebot`, `-t duckiedrone` selects the Duckiedrone type. `-c DD24` selects its compatible platform configuration and software stack.

The command asks you to enter and confirm a password for the virtual robot’s `duckie` account. Choose a personal password with at least eight characters. Enter it at the prompts; typed characters are not displayed.

The creation process might take a few minutes to prepare the software environment. Once created, starting the virtual robot, connecting to it, and doing [other supported operations](https://docs.duckietown.com/ente/duckietown-manual/50-duckiematrix/virtual-duckietown-robots/introduction-to-virtual-duckiebots.html), are separate steps:

```text
create → start → connect → work in the shell
                    ↑             |
                    +---- exit ---+
                         |
                   robot still running
```

## List, start, and connect

Once the creation process is complete, inspect the current status:

```bash
dts duckiebot virtual list
```

You should see a row for your each virtual Duckietown robot you have, including the recently created one. 

```text
         |    Type     | Model |    Status    | Uptime
-------- | ----------- | ----- | ------------ | ------
vamelia  | duckiedrone |  DD24 |     Down     |      -
vargo    |   duckiebot | DB21J |     Down     |      -
```

If it is not running (Status: Down), start it:

```bash
dts duckiebot virtual start vamelia
```

Then open its shell:

```bash
dts duckiebot virtual connect vamelia
```

This command checks that the virtual robot is running and opens an interactive Bash shell inside it. It does not use SSH or request the `duckie` password entered during creation. You will know the command succeeded if the prompt changes to indicate the virtual robot’s hostname:

```text
root@vamelia2:/#
```

Now establish where you are:

```bash
hostname
whoami
pwd
```

Read these results as three separate pieces of information:

| Command | What to check | Typical outcomes are: |
| --- | --- | --- |
| `hostname` | Which environment identifies itself as the current host? | `vamelia` (or the name you chose) |
| `whoami` | Which account is running the shell? | root |
| `pwd` | Which directory are you working in? | `/` (the root of the virtual robot’s filesystem) |

Try a familiar inspection command:

```bash
ls
```

It lists the current directory inside the virtual robot, and **not** your base station’s project directory.

You can return to the base station with:

```bash
exit
```

Run `hostname` again and compare it with the result from inside the virtual robot.

### Check your result

You should now be back in the base-station shell. Run:

```bash
dts duckiebot virtual list
```

The virtual Duckiedrone should still be running. `exit` closed your shell session, but it did not stop the environment.

If `connect` reports that the robot is not running, check its status and start the same named robot. You do not need to create another one.

## Stop a virtual Duckiedrone

When you have finished practicing, stop your virtual robot from the base-station terminal:

```bash
dts duckiebot virtual stop vamelia
dts duckiebot virtual list
```

Confirm that the named robot is no longer running.

Stopping preserves the virtual robot for later use. Next time, start it and connect again; creation is not a step you repeat for every session.

The [Duckiematrix command reference](https://docs.duckietown.com/ente/devmanual-duckiematrix/intermediate/virtual_duckietown_robots/dts-commands.html) documents this lifecycle.

## Local virtual and remote connections

Both workflows give you a shell in a robot environment, but they reach it differently:

| Physical Duckiedrone | Local virtual Duckiedrone |
| --- | --- |
| Connect with `ssh duckie@ROBOT_NAME.local` | Connect with `dts duckiebot virtual connect ROBOT_NAME` |
| Requires a network path to the robot | Uses the local virtual-robot environment |
| Checks the server’s identity and authenticates your account | Opens a shell through the local management command |
| `exit` closes the SSH session | `exit` closes the virtual-robot shell |

The virtual robot acts as a physical robot, and can still generate network traffic. Connections it makes to external services use the base station’s networking infrastructure and remain subject to the network’s access rules.
