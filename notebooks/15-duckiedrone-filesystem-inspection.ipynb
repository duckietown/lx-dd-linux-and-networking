# Duckiedrone Filesystem Inspection

You can now open a shell on a physical or virtual Duckiedrone. The next question is where to look: synchronized source code, Duckiedrone data, and Linux configuration occupy different parts of its filesystem. Some paths also provide interfaces to live system state rather than ordinary stored files.

This notebook applies the navigation and permission commands you already know to the Duckiedrone. You will locate the main directories, interpret a directory listing, and distinguish ordinary stored files from interfaces provided by the kernel.

Use an authorized Duckiedrone shell opened through SSH for a physical Duckiedrone or `dts duckiebot virtual connect` for a virtual one. The activities inspect directory names and metadata without changing files.

## Inspect filesystem landmarks

First, confirm your context and move to the filesystem root:

```bash
hostname
whoami
cd /
pwd
```

`hostname` identifies the environment, `whoami` identifies your account, and `pwd` should now print `/`.

Remember that `/` and `/root` are different: `/` is the top of the filesystem hierarchy; `/root` is the `root` account's home directory.

### Locate the Duckiedrone's code and data

Inspect these directories:

```bash
ls -ld /home/duckie /code /data
```

`-l` requests detailed information. `-d` lists each directory itself rather than its contents.

An illustrative line might look like:

```shell
drwxr-xr-x 2 root root 4096 Sep 25 10:00 /code
```

Read it using the permissions concepts from earlier notebooks:

- The initial `d` identifies a directory.

- `rwxr-xr-x` gives permissions for its owner, group, and others.

- `root root` names the owner and group.

- `/code` is the directory being described.

The size field is not the combined size of all files inside the directory. Your ownership, permissions, sizes, and dates may differ.

Now inspect their immediate contents:

```bash
ls -la /home/duckie
ls -la /code
ls -la /data
```

`-a` includes entries whose names begin with a dot and are typically hidden. These commands list entries without opening files, modifying them, or descending into subdirectories.

For example, if `/code` contains only `.` and `..`, it is an empty directory. If a path is missing or access is denied, record that result and continue with the other paths. Do not create directories or change permissions to make this example match.

### Recognize the Linux system directories

Duckiedrone-specific directories sit alongside the usual Linux structure:

```bash
ls -ld /etc /var /tmp
```

For example, `/tmp` can hold intermediate output during an exercise, but it is not a reliable place to keep a recording you need later. Temporary files may be cleaned up.

These conventional locations are described in the [Filesystem Hierarchy Standard](https://refspecs.linuxfoundation.org/FHS_3.0/fhs/index.html). Duckietown's `/code` and `/data` locations serve its own software workflow.

### Recognize kernel-provided interfaces

Finally, inspect:

```bash
ls -ld /proc /sys /dev
```

These paths expose information and interfaces rather than serving as ordinary project storage:

- `/proc` exposes process and kernel information. For example, `/proc/<PID>` describes a currently existing process with that identifier.

- `/sys` exposes the kernel's view of devices and their attributes.

- `/dev` contains device interfaces, such as terminal devices.

A file-like path in these locations can represent live system state or an interface to a device. For this tour, inspect only the directory metadata.

## Notice the user context

An SSH connection using `ssh duckie@...` enters the `duckie` account. The virtual connection workflow can enter as `root`, so check `whoami` rather than assuming the account from the Duckiedrone's name.

This is relevant when interpreting the same listing. In the illustrative `/code` permissions above, `root` has write permission, while the group and others do not. A command may therefore behave differently under the two accounts.

For this notebook, neither account needs to modify the filesystem. Leave `/secrets`, `/root`, and the contents of `/data/config` outside the inspection. Before sharing output or screenshots in reports, chat messages, or source files, review and redact sensitive values. Even directory names and metadata can reveal device-specific information.
